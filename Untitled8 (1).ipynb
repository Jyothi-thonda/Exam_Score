```python
df = df.drop_duplicates()

score_columns = [
    "math_score",
    "science_score",
    "english_score",
    "final_score"
]

for col in score_columns:
    df.loc[(df[col] < 0) | (df[col] > 100), col] = np.nan
```

```python
df.loc[df["study_hours"] < 0, "study_hours"] = np.nan
for col in df.columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicates:", df.duplicated().sum())

print("\nShape:", df.shape)
```

```python
df["average_subject_score"] = (
    df["math_score"] +
    df["science_score"] +
    df["english_score"]
) / 3

df["total_subject_score"] = (
    df["math_score"] +
    df["science_score"] +
    df["english_score"]
)
```

```python
df["study_intensity"] = df["study_hours"] / df["average_subject_score"]
```
